```python
# Сделайте функцию, которая параметром будет принимать число, а возвращать сумму его делителей.
def sum_of_divisors(n: int) -> int:
    if n <= 0:
        return 0
    if n == 1:
        return 1
    
    result = 0
    for i in range(1, n//2 + 1):
        if n % i == 0:
            result += i

    return result

print (sum_of_divisors(0))
print (sum_of_divisors(1))
print (sum_of_divisors(6))   # 1 + 2 + 3 = 6
print (sum_of_divisors(12))  # 1 + 2 + 3 + 4 + 6 = 16
print (sum_of_divisors(111)) # 1 + 3 + 37 = 41
```

```python
# Сделайте функцию, которая заполнит список N случайными числами из заданного промежутка.

import random
def generate_N_random(n: int, start:int, end:int) -> list:
    if (n < 1):
        return []

    return [random.randint(start, end) for x in range(n)]

print (generate_N_random(0, 1, 10))
print (generate_N_random(1, 1, 10))
print (generate_N_random(12, 1, 10))
```

```python
# Сделайте функцию, которая будет возвращать сумму N первых чисел Фибоначчи.

def fib_sum(n: int) -> int:
    #print (f"N={n} ", end="")
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:  
        fibs = [0, 1]
        for x in range(2,n):
            fibs.append(sum(fibs[-2:]))

        #print (fibs)
        return sum(fibs)


for i in range(1,10):
    print(fib_sum(i))

```

```python
#Сделайте функцию, которая параметром будет принимать секунды, а возвращать количество дней, часов, минут и секунд, соответствующих этим секундам, в виде следующего словаря:
#{
#	'd': 12,
#	'h': 10,
#	'm': 59,
#	's': 59,
#}

def seconds_to_dict(input: int) -> dict:
    days, remainder = divmod(input, 3600*24)
    hours, remainder = divmod(remainder, 3600)
    minutes, seconds = divmod(remainder, 60)

    return {
        'd': days,
        'h': hours,
        'm': minutes,
        's': seconds
    }

seconds_to_dict(3600*24*4 + 3600*12 + 60*5 + 34)
```

```python
# Сделайте функцию, которая параметром будет принимать список и элемент и возвращать следующий за ним элемент. Смотрите пример:
# lst = [1, 2, 3, 4, 5];
# func(lst, 1); // 2
# func(lst, 4); // 5
# func(lst, 5); // 1

def func(lst: list, n: int) -> int:
    if len(lst) == 0:
        return

    if n not in lst:
        return
    
    position = lst.index(n)
    return lst[(position + 1) % len(lst)]
    
print (func([],5))
print (func([1,2,3,4,5],9))
print (func([1,2,3,4,5],1))
print (func([1,2,3,4,5],4))
print (func([1,2,3,4,5],5))
```

```python
# Сделайте функцию, которая заполнит список N случайными числами из заданного промежутка так, чтобы числа не повторялись.
import random
def generate_N_uniq_random(n: int, output: list, start:int, end:int):
    if (n < 1):
        return []
    
    if (end - start + 1) < n:
        return []
 
    while len(output) < n:
        value = random.randint(start, end)
        if value not in output:
            output.append(value)

n_cases = [0, 3, 10, 12]
for n in n_cases:
    current_output = list()
    generate_N_uniq_random(n, current_output, 1, 10)
    print(f"N = {n}: {current_output}")
```

```python
# Сделайте функцию, которая параметром будет принимать двухмерный список чисел и возвращать список максимальных чисел в каждом подсписке.
def get_maximums(input: list) -> list:
    if len(input) < 1:
        return []

    result = list()
    for lst in input:
        result.append(max(lst))

    return result

input1 = [generate_N_uniq_random(9, 1, 10), generate_N_uniq_random(9, 1, 20), generate_N_uniq_random(9, 1, 100)]
print (get_maximums(input1))
```

```python
# Сделайте функцию, которая параметром будет принимать слово и возвращать список его слогов.

def split_by_syllable(word: str) -> list:
    word = word.lower()
    vowels = "аеёиоуыэюя"

    vowel_indices = [i for i, char in enumerate(word) if char in vowels]
    
    if len(vowel_indices) <= 1:
        return [word]

    result = list()
    position = 0

    for k in range(len(vowel_indices) - 1):
        curr_vowel_idx = vowel_indices[k]
        next_vowel_idx = vowel_indices[k+1]

        split_point = curr_vowel_idx + 1

        between = word[split_point : next_vowel_idx]
        # print (between, k, curr_vowel_idx)
        if len(between) > 1:
            first_consonant = between[0]
            second_consonant = between[1]

            if first_consonant == "й":
                split_point = curr_vowel_idx + 2
            elif first_consonant in "мнлр" and second_consonant not in "мнлрйъь":
                split_point = curr_vowel_idx + 2
    
        result.append(word[position:split_point])
        position = split_point

    result.append(word[position:])
    return result


words = ["упячка", "майка", "стойка", "смеркалось", "бабушка", "водка", "спектр", "шурф", "змееед", "застрахуй",
         "оскорблять"] # https://slogi.su/оскорблять

for w in words:
    print(f"{w} -> {split_by_syllable(w)}")
```
