```python
# Дан многомерный список произвольного уровня вложенности:
given_list = [
	{ 'a': [ { 'b': [ { 'c': '+++' } ] } ] }
]
# Сделайте функцию, которая будет возвращать элементы объекта, параметром получая строку с ключами объекта, разделенными точками:

def get_by_path(data, path: str) -> str:
    if not path:
        return ""
        
    parts = path.split('.', 1)
    current_key = parts[0]
    
    if isinstance(data, list):
        for item in data:
            result = get_by_path(item, path)
            if result is not None:
                return result
        return None

    if isinstance(data, dict) and current_key in data:
        if len(parts) == 1:
            return data[current_key]
        return get_by_path(data[current_key], parts[1])
        
    return None

print (get_by_path(given_list, 'a.b.c')) # ; // вернет '+++'
```

```python
#Дана строка, содержащая в себе любое количество вложенных парных круглых скобок. Напишите код, который проверит, что скобки расположены корректно.
# Так корректно:
# '()()'
# Так корректно:
# '(())'
# Так некорректно, так как количество открытых скобок не совпадает с количеством закрытых:
# '())'
# Так некорректно, так как, хотя количество открытых скобок и совпадает с количеством закрытых, но у них неправильный порядок:
# '))(('

def check_brackets(text: str) -> bool:
    counter = 0
    
    for char in text:
        if char == '(':
            counter += 1
        elif char == ')':
            counter -= 1
            
        if counter < 0:
            return False
            
    return (counter == 0)

# --- Проверка работы ---
print(check_brackets('()()'))
print(check_brackets('(())'))
print(check_brackets('())'))
print(check_brackets('))(('))
```
