```python
# Спросите у пользователя два целых числа. В ответ выведите список всех простых чисел из промежутка, заданном введенными значениями.
first=input()
second=input()

def is_prime(n: int) -> bool:
    for x in range(2, n//2 + 1):
        if n % x == 0:
            return False
    return True

if first.isdecimal() and second.isdecimal():
    start = int(first)
    end = int(second)
    
    if start > end:
        start, end = end, start

    prime_numbers = [num for num in range(start, end + 1) if is_prime(num)]
    print(prime_numbers)

```

```python
# Пусть у нас есть дата 22.02.2017. В этой дате 4 двойки. Найдите все даты с 4-мя двойками в текущем столетии.

from datetime import date, timedelta

# https://stackoverflow.com/questions/1060279/iterating-through-a-range-of-dates-in-python
def daterange(start_date: date, end_date: date):
    days = int((end_date - start_date).days)
    for n in range(days):
        yield (start_date + timedelta(n)).strftime("%Y-%m-%d")

start_date = date(2001, 1, 1)
end_date = date(2100, 12, 31)
dates = [d for d in daterange(start_date, end_date) if d.count('2') == 4]
print ('\n'.join(dates))

```

```python
#Дан некоторый URL:
url = 'http://test.com/dir1/dir2/dir3/page.html'
#Даны переменные:
num = 2;
val = 'eee'
#Замените в URL папку с номером num на значение val:
# 'http://test.com/dir1/eee/dir3/page.html'

def url_fixer(url: str, num: int, val: str) -> str:
    if (num > 0):
        parts = url.split('/')
        if (num <= len(parts) - 4):
            #print (parts, len(parts))
            domain_index = next(idx for idx, part in enumerate(parts) if '.' in part)
            parts[domain_index + num] = val
            result = '/'.join(parts)
            #print (result)
            return result

url_fixer(url,num,val)

```

```python
#Дан произвольный двухмерный список:
given_list = [
	[11, 12, 13, 14, 15],
	[21, 22, 23, 24, 25],
	[31, 32, 33, 34, 35],
	[41, 42, 43, 44, 45],
	[51, 52, 53, 54, 55],
]
#Переставьте местами два заданных столбца:
# [
# 	[11, 15, 13, 14, 12],
# 	[21, 25, 23, 24, 22],
# 	[31, 35, 33, 34, 32],
# 	[41, 45, 43, 44, 42],
# 	[51, 55, 53, 54, 52],
# ]

def swap_column(matrix:list, n: int, m:int):
    for row in matrix:
        row[n], row[m] = row[m], row[n]

swap_column(given_list, 1, 4)
print (given_list)
swap_column(given_list, 4, 1)
print (given_list)
```

```python
#Дан произвольный двухмерный список:
given_list = [
	[11, 32, 13, 44, 55],
	[31, 42, 53, 66, 75],
	[12, 26, 33, 45, 52],
	[15, 22, 35, 64, 57],
	[21, 52, 32, 44, 38],
]
# Найдите максимальный и минимальный элементы и поменяйте их местами.

def swap_min_max(matrix: list):
    min_r, min_c = 0, 0
    max_r, max_c = 0, 0

    for i in range(len(matrix)):
        for j in range(len(matrix[i])):
            if matrix[i][j] < matrix[min_r][min_c]:
                min_r, min_c = i, j
            if matrix[i][j] > matrix[max_r][max_c]:
                max_r, max_c = i, j

    matrix[min_r][min_c], matrix[max_r][max_c] = matrix[max_r][max_c], matrix[min_r][min_c]

print (given_list)
swap_min_max(given_list)
print (given_list)
```
