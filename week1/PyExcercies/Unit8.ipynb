```python
from itertools import chain
# Дан список городов и их стран, хранящийся в следующей структуре:
data = [
	{
		'country': 'country1',
		'city':    'city11',
	},
	{
		'country': 'country2',
		'city':    'city21',
	},
	{
		'country': 'country3',
		'city':    'city31',
	},
	{
		'country': 'country1',
		'city':    'city12',
	},
	{
		'country': 'country1',
		'city':    'city13',
	},
	{
		'country': 'country2',
		'city':    'city22',
	},
	{
		'country': 'country3',
		'city':    'city31',
	},
]

countries = list()

for elem in data:
    country = elem['country']
    city = elem['city']

    found = [d for d in countries if country in d]

    if not found:
        target_dict = {country: []}
        countries.append(target_dict)
    else:
        target_dict = found[0]
        
    if city not in target_dict[country]:
        target_dict[country].append(city)

print (countries)
```

```python
# Дан список событий за определенные месяцы, хранящийся в следующей структуре:
events = [
	{
		'date':  '2019-12',
		'event': 'name1'
	},
	{
		'date':  '2019-12',
		'event': 'name2'
	},
	{
		'date':  '2019-11',
		'event': 'name3'
	},
	{
		'date':  '2019-11',
		'event': 'name4'
	},
	{
		'date':  '2020-10',
		'event': 'name5'
	},
	{
		'date':  '2020-10',
		'event': 'name6'
	},
	{
		'date':  '2020-11',
		'event': 'name5'
	},
	{
		'date':  '2020-11',
		'event': 'name6'
	},
	{
		'date':  '2020-12',
		'event': 'name7'
	},
	{
		'date':  '2020-12',
		'event': 'name8'
	},
	{
		'date':  '2020-12',
		'event': 'name9'
	},
]
# Напишите код, которой переделает структуру данных вот в такую:

# {
# 	2019: {
# 		11: [список событий],
# 		12: [список событий],
# 	},
# 	2020: {
# 		10: [список событий],
# 		11: [список событий],
# 		12: [список событий],
# 	}
# }

years = dict();
for elem in events:
    year, month = map(int,elem['date'].split('-'))
    event = elem['event']
    if year not in years:
        years[year] = {month: [event]}
    else:
        if month not in years[year]:
            years[year][month] = [event]
        else:
            years[year][month].append(event)

#print (years)
sorted_years = {
    y: {m: years[y][m] for m in sorted(years[y])} 
    for y in sorted(years)
}
print (sorted_years)
```
